```python
import pandas as pd
from pathlib import Path

SOURCE_PATH = Path("../trace_annotation_log.xlsx")

df = pd.read_excel(SOURCE_PATH)

print(f"Total rows: {len(df)}")
print(f"Columns: {list(df.columns)}\n")
print("Verified Label distribution:")
print(df["Verified Label"].value_counts(dropna=False))

df.head()
```

```python
unsafe_df = df[df["Verified Label"] == "UNSAFE_EXECUTION"].copy()

print(f"UNSAFE_EXECUTION rows: {len(unsafe_df)}")
unsafe_df
```

```python
OUTPUT_PATH = Path("../unsafe_execution_cases.xlsx")
unsafe_df.to_excel(OUTPUT_PATH, index=False)
print(f"Saved {len(unsafe_df)} rows to {OUTPUT_PATH}")
```
