```python
import pandas as pd
from pathlib import Path

SOURCE_PATH = Path("../trace_annotation_log_backup_20260705_211956.xlsx")
OUTPUT_PATH = Path("../loop_verified_traces.xlsx")

df = pd.read_excel(SOURCE_PATH)
print(df.shape)
df["Verified Label"].value_counts()
```

```python
loop_df = df[df["Verified Label"] == "LOOP"].reset_index(drop=True)
print(f"Extracted {len(loop_df)} LOOP-verified rows")
loop_df
```

```python
loop_df.to_excel(OUTPUT_PATH, index=False)

check = pd.read_excel(OUTPUT_PATH)
assert len(check) == len(loop_df)
assert (check["Verified Label"] == "LOOP").all()
print(f"Saved {len(check)} rows to {OUTPUT_PATH.resolve()}")
```

```python
loop_df["Source"] = "REAL"
loop_df["Parent Trace ID"] = loop_df["Trace ID"]  # a real trace is its own parent

loop_df.to_excel(OUTPUT_PATH, index=False)
loop_df[["Trace ID", "Source", "Parent Trace ID"]]
```
